Everyone starts equally. Each round, everyone with money gives $1 to a randomly selected person. No debt. Predict the distribution before running.

Adapted from the course’s 2025 Mesa wealth notebook. Same sequential fixed-dollar rule; no Mesa installation needed. A person can select themselves (no change). This models random transfers, not production or payment for work.


```
import random
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output

def simulate(population=100, rounds=100, starting_money=1, seed=42):
    rng = random.Random(seed)
    wealth = [starting_money] * population
    for _ in range(rounds):
        # Sequential turns, as in the original notebook.
        for person in range(population):
            if wealth[person] > 0:
                recipient = rng.randrange(population)
                wealth[person] -= 1
                wealth[recipient] += 1
    return wealth

def plot_wealth(population=100, rounds=100, starting_money=1, seed=42):
    wealth = simulate(population, rounds, starting_money, seed)
    fig, ax = plt.subplots(figsize=(9, 4))
    maximum = max(wealth)
    if maximum <= 100:
        bins = [i - 0.5 for i in range(maximum + 2)]
    else:
        bins = 40
    ax.hist(wealth, bins=bins, color='#287da0', edgecolor='white')
    ax.set(xlabel='Wealth ($)', ylabel='Number of people',
           title=f'{population} people · {rounds} rounds · ${starting_money} each initially')
    fig.tight_layout()
    plt.show()
    print(f'Total: ${sum(wealth):,} | Mean: ${sum(wealth)/population:.2f}')
    print(f'At $0: {wealth.count(0)} of {population} ({wealth.count(0)/population:.1%})')
    print(f'Largest holding: ${maximum}')
    return wealth
```

## Run and compare
Change a setting, then click **Run**. Try 0, 10, 100, and 1,000 rounds. Keep the other settings fixed. Change the seed to repeat with different random draws.


```
population = widgets.IntSlider(value=100, min=10, max=2000, step=10, description='People', continuous_update=False)
rounds = widgets.IntSlider(value=100, min=0, max=2000, step=10, description='Rounds', continuous_update=False)
starting_money = widgets.IntSlider(value=1, min=1, max=100, description='Start ($)', continuous_update=False)
seed = widgets.IntText(value=42, description='Seed')
run = widgets.Button(description='Run', button_style='primary')
output = widgets.Output()

def run_clicked(_):
    run.disabled = True
    try:
        with output:
            clear_output(wait=True)
            plot_wealth(population.value, rounds.value, starting_money.value, seed.value)
    finally:
        run.disabled = False

run.on_click(run_clicked)
display(widgets.VBox([population, rounds, starting_money, seed, run, output]))
```

3. Can a person at $0 recover? How?

The standard fixed-dollar random-exchange model approaches a geometric distribution (an exponential shape for whole dollars), not a power law. Finite runs and update rules affect results; do not expect exact percentages. This histogram does not show whether the same people stay rich.

The yard-sale model uses a fraction of the poorer person’s wealth instead of a fixed dollar. Changing the exchange rule can change the collective outcome.
