## 1. Dictionary → DataFrame
Each dictionary key becomes a column. Each list supplies its values.

```python
import pandas as pd

data = {
    "department": ["Easy", "Easy", "Hard", "Hard"],
    "group": ["Men", "Women", "Men", "Women"],
    "applicants": [100, 10, 10, 100],
    "admitted": [80, 9, 1, 20]
}
print(data["applicants"])
admissions = pd.DataFrame(data, index=["easy_men", "easy_women", "hard_men", "hard_women"])
admissions
```

## 3. Calculate and plot
Admission rate = admitted ÷ applicants. Predict which group has the higher rate within each department.

```python
admissions["rate_percent"] = 100 * admissions["admitted"] / admissions["applicants"]
admissions
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))
plt.bar(admissions.index, admissions["rate_percent"])
plt.ylabel("Admitted (%)")
plt.ylim(0, 100)
plt.title("Fictional admissions: within departments")
plt.show()
```

## 4. Combine departments
Add admitted counts and applicant counts separately. Then divide. **Do not average the two department percentages.** Predict the combined result first.

```python
men_admitted = admissions.loc["easy_men", "admitted"] + admissions.loc["hard_men", "admitted"]
men_applicants = admissions.loc["easy_men", "applicants"] + admissions.loc["hard_men", "applicants"]
women_admitted = admissions.loc["easy_women", "admitted"] + admissions.loc["hard_women", "admitted"]
women_applicants = admissions.loc["easy_women", "applicants"] + admissions.loc["hard_women", "applicants"]

combined = pd.DataFrame({
    "applicants": [men_applicants, women_applicants],
    "admitted": [men_admitted, women_admitted]
}, index=["Men", "Women"])
combined["rate_percent"] = 100 * combined["admitted"] / combined["applicants"]
combined
```
