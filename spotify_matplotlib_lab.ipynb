## 1. Load the Spotify data
Run the setup cell. You do not need to understand every setup line yet.

```python
import pandas as pd
import matplotlib.pyplot as plt

DATA_URL = "https://huggingface.co/datasets/maharshipandya/spotify-tracks-dataset/resolve/main/dataset.csv"
raw_songs = pd.read_csv(DATA_URL)

songs = raw_songs[[
    "artists", "track_name", "popularity", "danceability",
    "energy", "valence", "acousticness", "tempo"
]].copy()
songs = songs.drop_duplicates(subset=["artists", "track_name"])
songs = songs[songs["popularity"] >= 50].copy()
songs = songs.rename(columns={"artists": "artist"})
songs.head()
```

### Meet the variables

- `danceability`: how suitable a track is for dancing
- `energy`: perceived intensity and activity
- `valence`: musical positivity
- `acousticness`: confidence that the track is acoustic
- `tempo`: estimated beats per minute

## 2. Build a first scatterplot
Run the cell, then identify what each line contributes to the finished visualization.

```python
plt.figure(figsize=(9, 6))
plt.scatter(songs["energy"], songs["valence"], alpha=0.15, color="steelblue")
plt.xlabel("Energy")
plt.ylabel("Valence (musical positivity)")
plt.title("Energy and Positivity in Popular Spotify Tracks")
plt.show()
```

### Show the same data as density

Now keep every song but group overlapping points into hexagons. The color shows how many tracks fall in each area.

```python
plt.figure(figsize=(9, 6))

plt.hexbin(
    songs["energy"],
    songs["valence"],
    gridsize=35,
    cmap="Blues",
    mincnt=1
)

plt.xlabel("Energy")
plt.ylabel("Positivity (Valence)")
plt.title("Energy and Positivity in Popular Spotify Tracks")
plt.colorbar(label="Number of tracks")
plt.show()
```

### What kinds of tracks occupy the lower-left corner?

The hexbin plot reveals a cluster of tracks with extremely low energy and low valence. Filter that corner of the graph, then count its most common genre labels to investigate what is creating the pattern.

```python
low_energy_low_valence = songs[
    (songs["energy"] < 0.05) &
    (songs["valence"] < 0.10)
]

genre_counts = (
    low_energy_low_valence["track_genre"]
    .value_counts()
    .head(12)
    .sort_values()
)

plt.figure(figsize=(10, 6))

genre_counts.plot(
    kind="barh",
    color="steelblue"
)

plt.title("Genres Among Very Low-Energy, Low-Positivity Tracks")
plt.xlabel("Number of tracks")
plt.ylabel("Genre")

plt.tight_layout()
plt.show()
```

### When the dots overlap: try a hexbin plot

A scatterplot can hide the pattern when thousands of points land on top of one another. A **hexbin plot** groups nearby songs into hexagons; darker hexagons contain more tracks. Run this version and compare it with the scatterplot above.

```python
plt.figure(figsize=(9, 6))

plt.hexbin(
    songs["danceability"],
    songs["acousticness"],
    gridsize=35,
    cmap="Purples",
    mincnt=1
)

plt.xlabel("Danceability")
plt.ylabel("Acousticness")
plt.title("Danceability and Acousticness in Popular Spotify Tracks")
plt.colorbar(label="Number of tracks")
plt.show()
```

## 4. Highlight a human category

Suppose we define a **high-energy heartbreak** song as energetic but low in valence. The thresholds below operationalize that human idea. Run the filter and inspect a few matches.

```python
heartbreak = songs[(songs["energy"] > 0.75) & (songs["valence"] < 0.30)]
heartbreak[["artist", "track_name", "energy", "valence"]].head(10)
```

```python
plt.figure(figsize=(9, 6))
plt.scatter(
    songs["energy"], songs["valence"],
    alpha=0.10, color="gray", label="all songs"
)
plt.scatter(
    heartbreak["energy"], heartbreak["valence"],
    alpha=0.55, color="crimson", label="high-energy heartbreak"
)
plt.xlabel("Energy")
plt.ylabel("Valence (musical positivity)")
plt.title("How We Turned ‘High-Energy Heartbreak’ into Data")
plt.legend()
plt.show()
```
