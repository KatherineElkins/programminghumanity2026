## Grammar and Sentence Generator

Run the next three cells to load a small context-free grammar and generate sentences from its rules.

**Questions:** Which generated sentences are grammatical but implausible? Why can the grammar generate them even when they do not make much sense?


```python
from nltk.parse.generate import generate, demo_grammar
from nltk import CFG

grammar = CFG.fromstring(demo_grammar)
print(grammar)
```

```python
for sentence in generate(grammar, n=5):
  print(' '.join(sentence))
```

```python
for sentence in generate(grammar, depth=10):
  print(' '.join(sentence))
```

## Ambiguous-Sentence Parser

Reference: [NLTK Chapter 8: Analyzing Sentence Structure](https://www.nltk.org/book/ch08.html)

Run the parser for *I shot an elephant in my pajamas*. Where does the phrase *in my pajamas* attach in each parse?


```python
import nltk
```

```python
groucho_grammar = nltk.CFG.fromstring("""
S -> NP VP
PP -> P NP
NP -> Det N | Det N PP | 'I'
VP -> V NP | VP PP
Det -> 'an' | 'my'
N -> 'elephant' | 'pajamas'
V -> 'shot'
P -> 'in'
""")
```

```python
sent = ['I', 'shot', 'an', 'elephant', 'in', 'my', 'pajamas']

parser = nltk.ChartParser(groucho_grammar)
for tree in parser.parse(sent):
  print(tree)
```
