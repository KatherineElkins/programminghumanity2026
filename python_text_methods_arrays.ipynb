## 4. List or array?
Use `type()` to check. `[2, 4, 1, 3]` creates a list; `np.array(...)` creates a NumPy array.

Lists can mix types. NumPy arrays use a common data type and support calculations across elements.


### Your answers
Predict what multiplying each object by 2 does.

Run it. Which repeats the list? Which doubles each value?

Change 2 to 3 and rerun. Which would you use to multiply word counts?


```python
import numpy as np

# Invented counts of one word in four sections of a text
counts_list = [2, 4, 1, 3]
counts_array = np.array(counts_list)

print("List type:", type(counts_list))
print("Array type:", type(counts_array))
print("List multiplied:", counts_list * 2)
print("Array multiplied:", counts_array * 2)
```

## Optional: counts versus rates
Which section has the most occurrences? Which has the most per 100 words? Predict, then run.


```python
import numpy as np

word_counts = np.array([2, 4, 1, 3])
section_lengths = np.array([100, 400, 100, 300])
per_100_words = word_counts / section_lengths * 100
print("Raw counts:", word_counts)
print("Occurrences per 100 words:", per_100_words)
```
