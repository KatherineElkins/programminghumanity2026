## 4. Square brackets: is it a list or an array?
A Python list and a NumPy array may look similar. **Use `type()` to check.** A list literal such as `[2, 4, 1, 3]` creates a list. `np.array(...)` creates a NumPy array. Importing NumPy as `np` gives the package a short name.

Lists can hold mixed types. A NumPy array has a common data type and supports numerical operations across its elements.


### Your answer 4 — identify and predict
**Before running:** Are the two objects the same type? Predict what multiplying each by 2 will do. [type here]

**After running:** Record both multiplication results. Explain which repeats a sequence and which calculates with each value. [type here]

**Change and test:** Change the multiplier from 2 to 3. Record what happens. Which object would you use to double every word count, and why? [type here]


```python
import numpy as np

# Invented counts of one word in four sections of a text
counts_list = [2, 4, 1, 3]
counts_array = np.array(counts_list)

print("List type:", type(counts_list))
print("Array type:", type(counts_array))
print("List multiplied:", counts_list * 2)
print("Array multiplied:", counts_array * 2)
```

## Optional challenge: raw counts and relative frequency
The sections below have different lengths. Predict which has the largest raw count and which has the highest rate per 100 words. Then run the code and explain the difference. Arrays let us calculate across all four sections in one operation.


```python
import numpy as np

word_counts = np.array([2, 4, 1, 3])
section_lengths = np.array([100, 400, 100, 300])
per_100_words = word_counts / section_lengths * 100
print("Raw counts:", word_counts)
print("Occurrences per 100 words:", per_100_words)
```
