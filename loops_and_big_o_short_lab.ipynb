## 1. A `for` loop: repeat an action for each item

Imagine digging a hole one inch at a time. Read the code in ordinary language:

> For each depth in the list, update the current depth.

Before running the cell, predict the final depth. Then add `6` and `7` to the list and run it again.


````python
depth_now = 0
depth_levels = [1, 2, 3, 4, 5]

for depth in depth_levels:
    depth_now = depth
    print("Now at", depth_now, "inches")

print("Finished at", depth_now, "inches")
````

**Discuss**

1. What action repeats?
2. What changes each time?
3. How does the loop know when to stop?


## 2. A `while` loop: repeat while a condition is true

Now imagine heating an oven. Read the code in ordinary language:

> While the oven is cooler than the target, raise the temperature.

Predict the final temperature. Then change the target to `400`.


````python
temperature = 70
target = 350

while temperature < target:
    temperature = temperature + 40
    print("Temperature:", temperature)

print("The oven is ready.")
````

**Discuss**

1. What condition controls the loop?
2. What value changes each time?
3. What would happen if the temperature never changed?


## 3. What Big-O means

Big-O does **not** tell us the exact number of seconds a program will take. It describes how the amount of work grows when the input grows.

| Growth | Ordinary-language description | Example |
|---|---|---|
| **O(1)** | Do one action, regardless of how many items exist | Read the first song in a playlist |
| **O(n)** | Do an action once for each item | Check every song to find the longest |
| **O(n²)** | For each item, go through all the items again | Compare every song with every other song |

The letter **n** means the number of items in the input.

## 5. Why loops matter for Big-O

- **One simple action** can be O(1).
- **One loop through n items** often produces O(n) work.
- **A loop inside another loop** often produces O(n²) work.

These are patterns, not rules that can be applied without examining what the code actually does.


````python
items = ["A", "B", "C"]

for item in items:
    print(item)
````

````python
items = ["A", "B", "C"]

for first_item in items:
    for second_item in items:
        print(first_item, second_item)
````

## 7. Recursion preview—no code to write

Iteration repeats an action through a loop. Recursion applies the same operation to a smaller instance or nested structure of the same kind until it reaches a **base case**.

```text
Count down from 3
└── Say 3, then count down from 2
    └── Say 2, then count down from 1
        └── Say 1, then count down from 0
            └── Base case: stop
```
