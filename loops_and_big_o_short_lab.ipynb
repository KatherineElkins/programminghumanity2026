## 2. A `while` loop repeats while a condition is true

Predict the output. Then change the target to `23`.

````python
target = 17
guess = 0
steps = 0

while guess != target:
    guess += 1
    steps += 1

print("Found", target, "in", steps, "steps")
````

## 3. O(1): the work stays approximately constant

This function retrieves one item. Predict whether a list with one million items requires more retrieval steps than a list with ten items.

````python
def take_first(values):
    steps = 1
    return values[0], steps

for size in [10, 100, 1_000, 10_000]:
    first, steps = take_first(list(range(size)))
    print(f"n = {size:>5,} | steps = {steps}")
````

## 4. O(n): the work grows with the input

Complete the missing line so the function finds the largest value without using `max()`.

````python
def find_largest(values):
    largest = values[0]
    steps = 0

    for value in values:
        steps += 1
        if value > largest:
            largest = value  # Replace or modify this line if needed.

    return largest, steps
````

````python
for size in [10, 100, 1_000, 10_000]:
    values = list(range(size))
    largest, steps = find_largest(values)
    print(f"n = {size:>5,} | steps = {steps:>5,}")
````

**Discuss:** When the input becomes ten times larger, what happens to the number of steps? Why is this O(n)?

## 5. O(n²): nested loops multiply the work

The function below compares every position with every other position. Predict the number of comparisons for 10 items before running it.

````python
def compare_every_pair(values):
    comparisons = 0

    for first in values:
        for second in values:
            comparisons += 1

    return comparisons

for size in [10, 20, 50, 100]:
    comparisons = compare_every_pair(range(size))
    print(f"n = {size:>3} | comparisons = {comparisons:>6,}")
````

**Discuss:**

1. Why does doubling the input produce approximately four times as many comparisons?
2. Where do you see a loop inside another loop?
3. Why might this pattern become slow even if it works well on a short list?
